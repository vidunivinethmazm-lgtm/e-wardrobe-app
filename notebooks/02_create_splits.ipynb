```python
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

CLEAN_CSV = BASE_DIR / "data" / "processed" / "clean_styles.csv"
SPLIT_DIR = BASE_DIR / "data" / "splits"
SPLIT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(CLEAN_CSV)

print("Dataset shape:", df.shape)
df.head()
```

```python
TARGET = "articleType"
MIN_SAMPLES = 500

class_counts = df[TARGET].value_counts()
valid_classes = class_counts[class_counts >= MIN_SAMPLES].index

df_filtered = df[df[TARGET].isin(valid_classes)].copy()

print("Original rows:", len(df))
print("Filtered rows:", len(df_filtered))
print("Number of classes:", df_filtered[TARGET].nunique())

df_filtered[TARGET].value_counts()
```

```python
train_df, temp_df = train_test_split(
    df_filtered,
    test_size=0.30,
    random_state=42,
    stratify=df_filtered[TARGET]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df[TARGET]
)

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

```python
print("Train class count:")
print(train_df[TARGET].value_counts())

print("\nValidation class count:")
print(val_df[TARGET].value_counts())

print("\nTest class count:")
print(test_df[TARGET].value_counts())
```

```python
train_df.to_csv(SPLIT_DIR / "train.csv", index=False)
val_df.to_csv(SPLIT_DIR / "val.csv", index=False)
test_df.to_csv(SPLIT_DIR / "test.csv", index=False)

print("Saved:")
print(SPLIT_DIR / "train.csv")
print(SPLIT_DIR / "val.csv")
print(SPLIT_DIR / "test.csv")
```
